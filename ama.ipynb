```python
import easyocr

# Create an EasyOCR reader
reader = easyocr.Reader(['en'])  # English language

# Read the image file
result = reader.readtext('image/hiwj3d4h.png')

# Print the extracted text
for (bbox, text, prob) in result:
    print("[INFO] {:.4f}: {}".format(prob, text))
```
