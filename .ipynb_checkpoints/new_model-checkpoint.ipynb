```python
import cv2
import easyocr
import subprocess
import matplotlib.pyplot as plt
import json
import re

def extract_text_from_image(image):
    """
    Extract text from the given image using EasyOCR.
    """
    try:
        reader = easyocr.Reader(['en'])
        result = reader.readtext(image, detail=0)
        return " ".join(result)
    except Exception as e:
        print(f"Error in EasyOCR: {e}")
        return ""

def clean_extracted_text(text):
    """
    Clean the extracted text by removing special characters and correcting OCR errors.
    """
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)  # Remove non-alphanumeric characters
    text = re.sub(r'\s+', ' ', text).strip()  # Remove extra spaces
    return text

def filter_extracted_text_with_llm(text):
    """
    Send extracted text to LLM to clean it by removing incomplete words, strange symbols, 
    and useless punctuation.
    """
    try:
        prompt = f"""
        Clean up the following text by removing incomplete words, special characters, and 
        any useless punctuation:
        "{text}"
        
        Return the cleaned text.
        """
        
        process = subprocess.run(
            ["ollama", "run", "gemma:2b"],  # Use the appropriate LLM command here
            input=prompt,
            text=True,
            capture_output=True,
            timeout=30
        )
        
        if process.returncode != 0:
            print(f"Ollama error: {process.stderr}")
            return text  # Return original text if filtering fails
            
        cleaned_output = process.stdout.strip()
        print(f"Filtered Text by LLM: {cleaned_output}")
        return cleaned_output
    except subprocess.TimeoutExpired:
        print("LLM query timed out.")
        return text  # Return original text if timeout occurs
    except Exception as e:
        print(f"Error during LLM filtering: {e}")
        return text  # Return original text if an exception occurs

def extract_product_name(text):
    """
    Extract potential product names by identifying fully capitalized words and limiting to 1-2 words.
    """
    capitalized_words = [word for word in text.split() if word.isupper()]
    
    # Limit product name to at most two words
    if len(capitalized_words) >= 2:
        product_name = " ".join(capitalized_words[:2])
    elif capitalized_words:
        product_name = capitalized_words[0]
    else:
        product_name = "Unknown Product"
    
    return product_name

def run_ollama_query(text, attempts=3):
    """
    Run Ollama query to extract product details from cleaned text multiple times.
    """
    try:
        product_name = extract_product_name(text)
        all_results = []

        for attempt in range(attempts):
            print(f"\nAttempt {attempt + 1}/{attempts} analyzing with Ollama...")
            prompt = f"""
            Extract product details from this text: {text}
            
            Use the capitalized words '{product_name}' as the product name, 
            and format the response as valid JSON with these fields:
            {{
                "ProductName": "{product_name}",
                "Brand": "brand name",
                "ProductType": "type of product",
                "ExtraInfo": "any additional details"
            }}
            """
            
            process = subprocess.run(
                ["ollama", "run", "gemma:2b"],  # Use the appropriate LLM command here
                input=prompt,
                text=True,
                capture_output=True,
                timeout=30
            )
            
            if process.returncode != 0:
                print(f"Ollama error: {process.stderr}")
                continue  # Skip to the next attempt if there's an error

            response = process.stdout.strip()
            print(f"Ollama output (attempt {attempt + 1}): {response}")
            
            try:
                json_start = response.find('{')
                json_end = response.rfind('}') + 1
                if json_start >= 0 and json_end > json_start:
                    json_str = response[json_start:json_end]
                    json_response = json.loads(json_str)
                    all_results.append(json_response)
                else:
                    all_results.append(response)
            except json.JSONDecodeError:
                print("Error: Response is not valid JSON.")
                all_results.append(response)

        # Aggregate the results (or return the last valid one)
        if all_results:
            return all_results
        else:
            return "No valid results from Ollama."
        
    except subprocess.TimeoutExpired:
        return "Ollama query timed out after 30 seconds."
    except Exception as e:
        print(f"Error in Ollama query: {e}")
        return f"Ollama call failed: {str(e)}"

def display_frame(frame, frame_count):
    """
    Display a single video frame using matplotlib.
    """
    plt.figure(figsize=(10, 8))
    plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    plt.title(f"Frame {frame_count}")
    plt.axis('off')
    plt.show(block=False)
    plt.pause(1)
    plt.close()

def process_video_for_product_details(video_path, frame_interval=30, num_frames_to_process=5):
    """
    Process a video file to extract product details by analyzing frames at specified intervals.
    """
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("Error: Could not open video.")
        return
    
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"Total frames in video: {total_frames}")
    
    frame_count = 0
    processed_frames = 0
    accumulated_text = set()

    try:
        while processed_frames < num_frames_to_process and frame_count < total_frames:
            ret, frame = cap.read()
            if not ret:
                print(f"Error reading frame {frame_count}")
                break

            if frame_count % frame_interval == 0:
                print(f"\nProcessing frame {frame_count}/{total_frames}...")

                # Resize and blur the frame for better OCR results
                frame = cv2.resize(frame, None, fx=1.5, fy=1.5)
                frame = cv2.GaussianBlur(frame, (3, 3), 0)
                
                # Display the frame
                display_frame(frame, frame_count)
                
                # Convert frame to RGB format for OCR
                rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                extracted_text = extract_text_from_image(rgb_frame)
                if extracted_text:
                    cleaned_text = clean_extracted_text(extracted_text)
                    accumulated_text.add(cleaned_text)
                    print(f"Extracted and cleaned text: {cleaned_text}")
                
                processed_frames += 1
            
            frame_count += 1

        if accumulated_text:
            print("\nAnalyzing accumulated text with LLM...")
            combined_text = " ".join(accumulated_text)
            
            # Step 1: Filter the text using the LLM
            filtered_text = filter_extracted_text_with_llm(combined_text)
            print(f"\nFiltered Text: {filtered_text}")
            
            # Step 2: Extract product information from filtered text (analyzing 3 times)
            product_details = run_ollama_query(filtered_text, attempts=3)
            print("\nExtracted Product Information (after 3 attempts):")
            for i, result in enumerate(product_details, 1):
                print(f"Attempt {i} Result:\n{json.dumps(result, indent=2) if isinstance(result, dict) else result}")
            
    finally:
        cap.release()
        plt.close('all')

if __name__ == "__main__":
    video_path = "panjabi.mp4"  # Replace with your video file path
    process_video_for_product_details(
        video_path,
        frame_interval=250,  # You can adjust the frame interval as needed
        num_frames_to_process=2  # Specify how many frames to process
    )
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing import image

# Load the saved model
model = tf.keras.models.load_model('fruit_freshness_model2.h5')

def predict_freshness(img_path):
    # Load and preprocess the image
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img) / 255.0  # Rescale
    img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension

    # Make prediction
    prediction = model.predict(img_array)
    print("fresness score",(1-prediction)*100)
    # Interpret the prediction
    if prediction[0] <= 0.6:
        return "Fresh"
    else:
        return "Rotten"

# Example usage
img_path = 'dataset/fresh/fresh_fruit_2.jpg'  # Path to the image you want to classify
result = predict_freshness(img_path)
print(f"The fruit is: {result}")
```
