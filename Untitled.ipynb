````python
import cv2
import easyocr
import subprocess
import matplotlib.pyplot as plt


# Function to extract text from an image using EasyOCR
def extract_text_from_image(image):
    try:
        reader = easyocr.Reader(['en'])  # Specify the language of the text
        result = reader.readtext(image, detail=0)  # detail=0 returns just the text
        print(f"Extracted Text: {result}")  # Debug: print the extracted text
        return " ".join(result)  # Combine the list of text into a single string
    except Exception as e:
        print(f"Error in EasyOCR: {e}")
        return ""


# Function to interactively send text to the Gemma 2B (Ollama) model
def get_product_details(text):
    try:
        # The prompt to be sent to the model
        prompt = f"""
        Extract the product details from the text delimited by triple backticks.
        Provide the details in JSON format:

        ``` 
        {{
          "ProductDetails": {{
            "ProductName": "",
            "ProductType": "",
            "ExpiryDate": "",
            "ManufactureDate": "",
            "OtherRelevantDates": []
          }}
        }}
        ```

        Text: ```{text}```
        """

        # Start the Gemma model using subprocess and send the prompt
        process = subprocess.Popen(["ollama", "run", "gemma:2b"], stdin=subprocess.PIPE, stdout=subprocess.PIPE,
                                   stderr=subprocess.PIPE, text=True)

        # Pass the prompt to the model
        output, error = process.communicate(input=prompt)

        # If there's any error during processing, print it
        if error:
            print(f"Error in Ollama API Call: {error}")
            return "Ollama call failed."

        print(f"Ollama Output: {output.strip()}")  # Debug: print the response from Ollama
        return output.strip()

    except Exception as e:
        print(f"Error in Ollama API Call: {e}")
        return "Ollama call failed."


# Function to display the current frame using matplotlib
def display_frame(frame, frame_count):
    plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))  # Convert from BGR to RGB for correct display
    plt.title(f"Frame {frame_count}")
    plt.axis('off')  # Hide axis
    plt.show()  # Display the frame


# Function to process multiple images (frames) and accumulate text for Gemma analysis
def process_video_for_product_details(video_path, frame_interval=30, num_frames_to_process=5):
    cap = cv2.VideoCapture(video_path)  # Open video file
    frame_count = 0
    accumulated_text = ""  # Store accumulated text from multiple frames

    if not cap.isOpened():
        print("Error: Could not open video.")
        return

    processed_frames = 0

    while cap.isOpened() and processed_frames < num_frames_to_process:
        ret, frame = cap.read()  # Capture frame-by-frame
        if not ret:
            print(f"Error reading frame {frame_count}")
            break

        # Process every Nth frame (based on the frame_interval)
        if frame_count % frame_interval == 0:
            print(f"Processing frame {frame_count}...")

            # Display the current frame
            display_frame(frame, frame_count)  # Use matplotlib to display the frame

            # Convert the frame to RGB (from BGR, which OpenCV uses)
            rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

            # Step 1: Extract text from the frame
            extracted_text = extract_text_from_image(rgb_frame)
            print(f"Extracted text from frame {frame_count}: {extracted_text}")

            # Step 2: Accumulate the extracted text
            accumulated_text += f" {extracted_text}"

            # Increase the processed frame count
            processed_frames += 1

        frame_count += 1

    # Step 3: Use Gemma 2B to get product details from the accumulated text
    if accumulated_text:  # Proceed only if some text was found
        print("Sending accumulated text to Gemma 2B for analysis...")
        product_details = get_product_details(accumulated_text)
        print("Extracted Product Information from Multiple Frames:")
        print(product_details)

    cap.release()


if __name__ == "__main__":
    video_path = "panjabi.mp4"  # Replace with your video file path
    process_video_for_product_details(video_path, frame_interval=100,
                                      num_frames_to_process=5)  # Process every 100th frame and accumulate data from 5 frames
````

````python
import cv2
import easyocr
import subprocess
import matplotlib.pyplot as plt
import json
import time

def extract_text_from_image(image):
    try:
        reader = easyocr.Reader(['en'])
        result = reader.readtext(image, detail=0)
        return " ".join(result)
    except Exception as e:
        print(f"Error in EasyOCR: {e}")
        return ""

def init_ollama_gemma():
    try:
        # Add timeout and shell=True for Windows compatibility
        process = subprocess.Popen(
            ["ollama", "run", "gemma:2b"],
            stdin=subprocess.PIPE,
            stdout=subprocess.PIPE,
            stderr=subprocess.PIPE,
            text=True,
            shell=True
        )
        
        # Wait briefly to ensure the model is loaded
        time.sleep(2)
        return process
    except Exception as e:
        print(f"Error starting Ollama Gemma: {e}")
        return None

def get_product_details(text, process):
    try:
        # Simplified and more focused prompt
        prompt = f"""
        Extract product details from this text: {text}
        
        Format the response as valid JSON with these fields:
        {{
            "ProductName": "full product name",
            "Brand": "brand name",
            "ProductType": "type of product",
            "ExtraInfo": "any additional details"
        }}
        """
        
        # Write prompt to stdin and flush
        process.stdin.write(prompt + "\n")
        process.stdin.flush()
        
        # Read response with timeout
        response = ""
        start_time = time.time()
        while time.time() - start_time < 10:  # 10 second timeout
            if process.stdout.readable():
                line = process.stdout.readline()
                if line.strip():
                    response += line
                if "}" in line:  # Basic check for JSON completion
                    break
            time.sleep(0.1)
        
        try:
            # Try to parse and format the JSON response
            json_response = json.loads(response)
            return json.dumps(json_response, indent=2)
        except json.JSONDecodeError:
            return response.strip()
            
    except Exception as e:
        print(f"Error in Ollama API Call: {e}")
        return "Ollama call failed."

def display_frame(frame, frame_count):
    plt.figure(figsize=(10, 8))
    plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    plt.title(f"Frame {frame_count}")
    plt.axis('off')
    plt.show(block=False)
    plt.pause(1)
    plt.close()

def process_video_for_product_details(video_path, frame_interval=30, num_frames_to_process=5):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("Error: Could not open video.")
        return
    
    # Get video properties
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"Total frames in video: {total_frames}")
    
    frame_count = 0
    processed_frames = 0
    accumulated_text = set()  # Use set to avoid duplicates
    
    # Initialize Ollama
    print("Initializing Ollama Gemma model...")
    ollama_process = init_ollama_gemma()
    if not ollama_process:
        print("Error: Ollama Gemma model could not be started.")
        return

    try:
        while processed_frames < num_frames_to_process and frame_count < total_frames:
            ret, frame = cap.read()
            if not ret:
                print(f"Error reading frame {frame_count}")
                break

            if frame_count % frame_interval == 0:
                print(f"\nProcessing frame {frame_count}/{total_frames}...")
                
                # Preprocess frame
                frame = cv2.resize(frame, None, fx=1.5, fy=1.5)  # Upscale for better OCR
                frame = cv2.GaussianBlur(frame, (3,3), 0)  # Reduce noise
                
                # Display frame
                display_frame(frame, frame_count)
                
                # Extract and accumulate text
                rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                extracted_text = extract_text_from_image(rgb_frame)
                if extracted_text:
                    accumulated_text.add(extracted_text)
                    print(f"Extracted text: {extracted_text}")
                
                processed_frames += 1
            
            frame_count += 1

        # Analyze accumulated text
        if accumulated_text:
            print("\nAnalyzing accumulated text with Gemma...")
            combined_text = " ".join(accumulated_text)
            product_details = get_product_details(combined_text, ollama_process)
            print("\nExtracted Product Information:")
            print(product_details)
            
    finally:
        cap.release()
        if ollama_process:
            ollama_process.terminate()
        plt.close('all')

if __name__ == "__main__":
    video_path = "panjabi.mp4"
    process_video_for_product_details(
        video_path,
        frame_interval=100,
        num_frames_to_process=5
    )
````

````python
import cv2
import easyocr
import subprocess
import matplotlib.pyplot as plt
import json
import time
from typing import Optional, Dict, Any

def extract_text_from_image(image):
    try:
        reader = easyocr.Reader(['en'])
        result = reader.readtext(image, detail=0)
        return " ".join(result)
    except Exception as e:
        print(f"Error in EasyOCR: {e}")
        return ""

def run_llama_query(text, max_retries=3) -> Optional[Dict[Any, Any]]:
    """
    Run Llama2-7b query with retries and shorter timeout
    """
    for attempt in range(max_retries):
        try:
            # Shorter, more focused prompt
            prompt = f"""
            Extract product details from this text and return only a JSON object:
            Text: {text}
            Expected format:
            {{
                "ProductName": "name",
                "Brand": "brand",
                "ProductType": "type",
                "ExtraInfo": "details"
            }}
            """
            
            # Run Llama2-7b with shorter timeout
            process = subprocess.run(
                ["ollama", "run", "llama2:7b"],
                input=prompt,
                text=True,
                capture_output=True,
                timeout=20  # Shorter timeout
            )
            
            if process.returncode != 0:
                print(f"Attempt {attempt + 1}: Llama2 error: {process.stderr}")
                continue
                
            response = process.stdout.strip()
            
            # Find and extract JSON content
            json_start = response.find('{')
            json_end = response.rfind('}') + 1
            
            if json_start >= 0 and json_end > json_start:
                json_str = response[json_start:json_end]
                # Clean the JSON string
                json_str = (json_str.replace('\n', ' ')
                           .replace('```', '')
                           .replace('json', ''))
                
                try:
                    return json.loads(json_str)
                except json.JSONDecodeError as je:
                    print(f"Attempt {attempt + 1}: JSON parsing error: {je}")
                    continue
            
            print(f"Attempt {attempt + 1}: No valid JSON found in response")
            
        except subprocess.TimeoutExpired:
            print(f"Attempt {attempt + 1}: Query timed out after 20 seconds")
        except Exception as e:
            print(f"Attempt {attempt + 1}: Error: {e}")
        
        # Wait before retrying
        if attempt < max_retries - 1:
            time.sleep(2)
    
    return None

def display_frame(frame, frame_count):
    plt.figure(figsize=(10, 8))
    plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    plt.title(f"Frame {frame_count}")
    plt.axis('off')
    plt.show(block=False)
    plt.pause(1)
    plt.close()

def preprocess_frame(frame):
    """
    Enhanced frame preprocessing for better OCR
    """
    try:
        # Convert to grayscale
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        
        # Apply adaptive thresholding
        thresh = cv2.adaptiveThreshold(
            gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, 
            cv2.THRESH_BINARY, 11, 2
        )
        
        # Denoise
        denoised = cv2.fastNlMeansDenoising(thresh)
        
        # Resize (upscale)
        resized = cv2.resize(denoised, None, fx=1.5, fy=1.5, 
                           interpolation=cv2.INTER_CUBIC)
        
        return resized
    except Exception as e:
        print(f"Error in frame preprocessing: {e}")
        return frame

def process_video_for_product_details(video_path, frame_interval=30, num_frames_to_process=5):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("Error: Could not open video.")
        return
    
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"Total frames in video: {total_frames}")
    
    frame_count = 0
    processed_frames = 0
    best_text = ""
    max_text_length = 0

    try:
        while processed_frames < num_frames_to_process and frame_count < total_frames:
            ret, frame = cap.read()
            if not ret:
                break

            if frame_count % frame_interval == 0:
                print(f"\nProcessing frame {frame_count}/{total_frames}...")
                
                # Preprocess frame
                processed_frame = preprocess_frame(frame)
                
                # Display original frame
                display_frame(frame, frame_count)
                
                # Extract text
                extracted_text = extract_text_from_image(processed_frame)
                
                # Keep the frame with the most text
                if len(extracted_text) > max_text_length:
                    max_text_length = len(extracted_text)
                    best_text = extracted_text
                
                if extracted_text:
                    print(f"Extracted text: {extracted_text}")
                
                processed_frames += 1
            
            frame_count += 1

        # Analyze best text
        if best_text:
            print("\nAnalyzing text with Llama2-7b...")
            product_details = run_llama_query(best_text)
            
            if product_details:
                print("\nExtracted Product Information:")
                print(json.dumps(product_details, indent=2))
            else:
                print("\nFailed to extract product information after all retries")
            
    finally:
        cap.release()
        plt.close('all')

if __name__ == "__main__":
    video_path = "panjabi.mp4"
    process_video_for_product_details(
        video_path,
        frame_interval=100,
        num_frames_to_process=5
    )
````

````python
import cv2
import easyocr
import subprocess
import matplotlib.pyplot as plt
import json
import re

def extract_text_from_image(image):
    """
    Extract text from the given image using EasyOCR.
    """
    try:
        reader = easyocr.Reader(['en'])
        result = reader.readtext(image, detail=0)
        return " ".join(result)
    except Exception as e:
        print(f"Error in EasyOCR: {e}")
        return ""

def clean_extracted_text(text):
    """
    Clean the extracted text by removing special characters and correcting OCR errors.
    """
    # More robust cleaning
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    text = ' '.join(word for word in text.split() if len(word) > 1)  # Remove single characters
    return text.strip()

def extract_product_name(text):
    """
    Extract potential product names using pattern matching and capitalization.
    """
    words = text.split()
    
    # Find sequences of capitalized words
    product_words = []
    current_sequence = []
    
    for word in words:
        if word.isupper() and len(word) > 1:
            current_sequence.append(word)
        else:
            if len(current_sequence) > 0:
                product_words.extend(current_sequence)
                current_sequence = []
    
    if current_sequence:  # Add any remaining sequence
        product_words.extend(current_sequence)
    
    # Take up to 3 consecutive capitalized words
    if product_words:
        return " ".join(product_words[:3])
    return "Unknown Product"

def run_llm_query(prompt, max_retries=3, timeout=45):
    """
    Run LLM query with retry mechanism and proper error handling.
    """
    for attempt in range(max_retries):
        try:
            process = subprocess.run(
                ["ollama", "run", "llama2"],
                input=prompt,
                text=True,
                capture_output=True,
                timeout=timeout
            )
            
            if process.returncode == 0:
                return process.stdout.strip()
            else:
                print(f"Attempt {attempt + 1} failed: {process.stderr}")
                
        except subprocess.TimeoutExpired:
            print(f"Attempt {attempt + 1} timed out after {timeout} seconds")
        except Exception as e:
            print(f"Attempt {attempt + 1} failed with error: {str(e)}")
            
        if attempt < max_retries - 1:
            print("Retrying...")
    
    return None

def filter_extracted_text_with_llm(text):
    """
    Filter and clean extracted text using LLM.
    """
    prompt = f"""
    Analyze this OCR extracted text and return only relevant product information.
    Ignore manufacturing details, barcodes, and addresses.
    Text: "{text}"
    Keep only product name, brand, type, and key features.
    """
    
    result = run_llm_query(prompt)
    return result if result else text

def parse_json_response(response):
    """
    Safely parse JSON from LLM response with fallback handling.
    """
    if not response:
        return None
        
    # Try to find JSON pattern
    json_match = re.search(r'\{[^{]*"ProductName".*?\}', response, re.DOTALL)
    if json_match:
        try:
            return json.loads(json_match.group(0))
        except json.JSONDecodeError:
            pass
    
    # Fallback: Try to extract structured information
    product_info = {
        "ProductName": "Unknown",
        "Brand": "Unknown",
        "ProductType": "Unknown",
        "ExtraInfo": ""
    }
    
    # Extract what we can from the response
    lines = response.split('\n')
    for line in lines:
        line = line.strip()
        if ':' in line:
            key, value = line.split(':', 1)
            key = key.strip()
            value = value.strip()
            
            if 'product' in key.lower() and len(value) > 1:
                product_info["ProductName"] = value
            elif 'brand' in key.lower() and len(value) > 1:
                product_info["Brand"] = value
            elif 'type' in key.lower() and len(value) > 1:
                product_info["ProductType"] = value
    
    return product_info

def run_ollama_query(text, attempts=2):
    """
    Extract product details from text using LLM.
    """
    product_name = extract_product_name(text)
    results = []
    
    prompt = f"""
    Analyze this product text and extract details in JSON format:
    Text: {text}
    
    Return a JSON object with these fields:
    {{
        "ProductName": "{product_name}",
        "Brand": "detected brand name",
        "ProductType": "type of product",
        "ExtraInfo": "key features or details"
    }}
    
    Respond ONLY with the JSON object, no other text.
    """
    
    for attempt in range(attempts):
        response = run_llm_query(prompt)
        if response:
            parsed_result = parse_json_response(response)
            if parsed_result:
                results.append(parsed_result)
    
    if not results:
        # Fallback result
        return [{
            "ProductName": product_name,
            "Brand": "Unknown",
            "ProductType": "Unknown",
            "ExtraInfo": "Failed to extract details"
        }]
    
    return results

def process_video_for_product_details(video_path, frame_interval=30, num_frames_to_process=5):
    """
    Process video frames to extract product information.
    """
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("Error: Could not open video.")
        return
    
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"Total frames in video: {total_frames}")
    
    frame_count = 0
    processed_frames = 0
    accumulated_text = set()
    
    try:
        while processed_frames < num_frames_to_process and frame_count < total_frames:
            ret, frame = cap.read()
            if not ret:
                break

            if frame_count % frame_interval == 0:
                print(f"\nProcessing frame {frame_count}/{total_frames}...")
                
                # Enhance frame for better OCR
                frame = cv2.resize(frame, None, fx=1.5, fy=1.5)
                frame = cv2.GaussianBlur(frame, (3, 3), 0)
                
                # Display frame
                display_frame(frame, frame_count)
                
                # Extract and clean text
                rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                extracted_text = extract_text_from_image(rgb_frame)
                if extracted_text:
                    cleaned_text = clean_extracted_text(extracted_text)
                    if cleaned_text:
                        accumulated_text.add(cleaned_text)
                        print(f"Extracted text: {cleaned_text}")
                
                processed_frames += 1
            frame_count += 1

        if accumulated_text:
            combined_text = " ".join(accumulated_text)
            filtered_text = filter_extracted_text_with_llm(combined_text)
            
            if filtered_text:
                print("\nAnalyzing product details...")
                results = run_ollama_query(filtered_text, attempts=2)
                
                print("\nExtracted Product Information:")
                for i, result in enumerate(results, 1):
                    print(f"\nResult {i}:")
                    print(json.dumps(result, indent=2))
                
                return results
            
    except Exception as e:
        print(f"Error processing video: {str(e)}")
    finally:
        cap.release()
        plt.close('all')

def display_frame(frame, frame_count):
    """
    Display video frame using matplotlib.
    """
    plt.figure(figsize=(10, 8))
    plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    plt.title(f"Frame {frame_count}")
    plt.axis('off')
    plt.show(block=False)
    plt.pause(1)
    plt.close()

if __name__ == "__main__":
    video_path = "panjabi.mp4"
    results = process_video_for_product_details(
        video_path,
        frame_interval=250,
        num_frames_to_process=2
    )
````
