```python
import easyocr
import cv2
import matplotlib.pyplot as plt

def preprocess_image(image_path):
    # Load image in grayscale
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    
    # Apply some preprocessing (e.g., thresholding)
    _, img_bin = cv2.threshold(img, 150, 255, cv2.THRESH_BINARY)
    
    return img_bin

def extract_text_easyocr(image_path):
    # Preprocess the image
    preprocessed_image = preprocess_image(image_path)
    
    # Initialize the EasyOCR reader
    reader = easyocr.Reader(['en'])  # Add more languages if needed
    
    # Extract text from the NumPy array directly
    result = reader.readtext(preprocessed_image)
    
    # Extract text and return it
    extracted_text = " ".join([res[1] for res in result])
    return extracted_text

# Test the function
image_path = 'dse9ol7h.png'  # Replace with your image file path
extracted_text = extract_text_easyocr(image_path)

print("Extracted Text:")
print(extracted_text)

# Display the image (optional)
plt.imshow(cv2.imread(image_path))
plt.title("Input Image")
plt.show()
```
